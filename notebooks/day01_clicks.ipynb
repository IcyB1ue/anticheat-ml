# First Notebook and Anti-Cheat Detection
This Notebook serves as an introduction to using ipynb files, along with an example of catching an autoclicker player.

```python
import numpy as np
player_a = np.array([0.00, 0.13, 0.24, 0.41, 0.52, 0.66, 0.75, 0.93, 1.04, 1.16])
player_b = np.array([0.00, 0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90])
```

```python
np.diff(player_a)
```

```python
np.diff(player_b)
```

```python
gaps_a = len(player_a) - 1
gaps_b = len(player_b) - 1

dt_a = player_a[-1] - player_a[0]
dt_b = player_b[-1] - player_b[0]

cps_a = gaps_a / dt_a
cps_b = gaps_b / dt_b

print(f"The Average CPS of Player A is {cps_a:.2f}")
print(f"The Average CPS of Player B is {cps_b:.2f}")
```

```python
print(np.std(np.diff(player_a))) # Standard Deviation of the Gaps for Player A
print(np.std(np.diff(player_b))) # Standard Deviation of the Gaps for Player B
```

The Autoclicker is Player B because it shouldn't be humanly possible to be that consistent with the timing of your clicks.
